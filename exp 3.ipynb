```python
# Import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the image in grayscale format
img = cv2.imread('lion.jpg', cv2.IMREAD_GRAYSCALE)
```

```python
# Display the grayscale image.
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.show()
```

```python
# Plot the histogram of the grayscale image

plt.hist(img.ravel(),256,range = [0, 256]);
plt.title('Original Image')
plt.show()
```

```python
# Equalize histogram
img_eq = cv2.equalizeHist(img)
```

```python
# Display the images.
plt.hist(img_eq.ravel(), 256, range = [0, 256])
plt.title('Equalized Histogram')
```

```python
# Display the images.
plt.imshow(img_eq, cmap='gray')
plt.title('Original Image')
plt.show()
```

```python
Write a Python program using OpenCV to perform color image enhancement through histogram equalization in the HSV color space on the given image

The program should:
    •	Import the required libraries (OpenCV, NumPy, Matplotlib).
    •	Read the color image "parrot.jpg".
    •	Plot the histogram of the original color image (use the B, G, and R channels).
    •	Convert the image from BGR to HSV color space.
    •	Apply histogram equalization to one specific channel to improve brightness and contrast using cv2.equalizeHist().
    •	Convert the enhanced HSV image back to BGR format.
    •	Using Matplotlib, display the following in a 2 × 2 grid layout:
            1.	Original Color Image
            2.	Histogram of Original Image (BGR channels)
            3.	Enhanced Color Image
            4.	Histogram of Enhanced Image (BGR channels)

```

```python
# Import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the colorgiven parrot.jpg image.
img = cv2.imread('lion.jpg', cv2.IMREAD_COLOR)

# Convert to HSV.
img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Perform histogram equalization only on the V channel, for value intensity.
img_hsv[:,:,2] = cv2.equalizeHist(img_hsv[:, :, 2])

# Convert back to BGR format.
img_eq = cv2.cvtColor(img_hsv, cv2.COLOR_HSV2BGR)
```

```python
plt.imshow(img_eq[:,:,::-1])
plt.title('Equalized Image')
plt.show()
```

```python
plt.hist(img_eq.ravel(),256,range = [0, 256])
plt.title('Histogram Equalized')
plt.show()
```

```python
# Better and cleaner visualization
plt.figure(figsize=(12, 8))

plt.subplot(221)
plt.imshow(img[:, :, ::-1])
plt.title('Original Color Image')

plt.subplot(222)
plt.imshow(img_eq[:, :, ::-1])
plt.title('Equalized Image')

plt.subplot(223)
plt.hist(img.ravel(),256,range = [0, 256])
plt.title('Original Image')

plt.subplot(224)
plt.hist(img_eq.ravel(),256,range = [0, 256])
plt.title('Histogram Equalized')
plt.show()


plt.tight_layout(pad=3)
plt.show()
```

```python
# Display the histograms.
plt.figure(figsize = [15,4])
plt.subplot(121)
plt.hist(img.ravel(),256,range = [0, 256])
plt.title('Original Image')

plt.subplot(122)
plt.hist(img_eq.ravel(),256,range = [0, 256])
plt.title('Histogram Equalized')
```
